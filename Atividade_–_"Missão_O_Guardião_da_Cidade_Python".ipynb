```python
# ==========================================
# MISSÃO: O GUARDIÃO DA CIDADE PYTHON
# ==========================================

def fase_1():
    print("--- FASE 1: Criando o Personagem ---")
    nome = input("Digite o seu nome: ")
    try:
        idade = int(input("Digite a sua idade: "))
    except ValueError:
        print("\nEntrada inválida! Digite apenas números para a idade.")
        print("Jogo encerrado.")
        return None

    if idade < 12:
        print("\nVocê é muito jovem para esta missão.")
        print("Jogo encerrado.")
        return None
    elif idade <= 17:
        print(f"\nSeja bem-vindo, {nome}! Você será um aprendiz.")
    else:
        print(f"\nSeja bem-vindo, {nome}! Você será um guerreiro programador.")
    return nome


def fase_2():
    print("\n--- FASE 2: A Porta do Conhecimento ---")
    print("Qual linguagem estamos aprendendo?")
    print("1 - Python")
    print("2 - Java")
    print("3 - PHP")
    try:
        opcao = int(input("Escolha uma opção (1, 2 ou 3): "))
    except ValueError:
        opcao = -1

    if opcao == 1:
        print("Resposta correta! Você ganhou 20 pontos.")
        return 20
    elif opcao == 2:
        print("Você escolheu Java e perdeu 5 pontos.")
        return -5
    elif opcao == 3:
        print("Você escolheu PHP e perdeu 10 pontos.")
        return -10
    else:
        print("Opção inválida! Você não ganhou nem perdeu pontos.")
        return 0


def fase_3():
    print("\n--- FASE 3: A Floresta dos Desafios ---")
    print("Você chegou a uma floresta e precisa escolher um caminho:")
    print("1 - Caminho da Lógica")
    print("2 - Caminho dos Bugs")
    print("3 - Caminho da Interface")
    try:
        opcao = int(input("Escolha o seu caminho (1, 2 ou 3): "))
    except ValueError:
        opcao = -1

    if opcao == 1:
        print("\nVocê resolveu um algoritmo!")
        return 30
    elif opcao == 2:
        print("\nVocê corrigiu alguns erros!")
        return 10
    elif opcao == 3:
        print("\nVocê criou uma tela bonita!")
        return 20
    else:
        print("\nOpção inválida! Você se perdeu na floresta e não pontuou.")
        return 0


def fase_4(nome, pontuacao):
    print("\n--- FASE 4: O Desafio Final ---")
    print(f"Jogador: {nome}")
    print(f"Pontuação Final: {pontuacao} pontos")

    if pontuacao >= 50:
        print("Resultado: Mestre Python!")
    elif pontuacao >= 30 and pontuacao <= 49:
        print("Resultado: Programador Aprendiz!")
    else:
        print("Resultado: Precisa treinar!")


# ------------------------------------------
# FLUXO PRINCIPAL DO JOGO
# ------------------------------------------
def iniciar_jogo():
    print("=" * 50)
    print("MISSÃO: O GUARDIÃO DA CIDADE PYTHON")
    print("=" * 50)
    print("Prepare-se para iniciar a sua jornada!\n")

    pontuacao = 0

    # Fase 1
    nome = fase_1()
    if nome is None:
        return  # Encerra o jogo se for menor de 12 ou digitar texto na idade

    # Fase 2
    pontuacao += fase_2()
    print(f"Pontuação atual: {pontuacao} pontos.")

    # Fase 3
    pontuacao += fase_3()
    print(f"Pontuação atual: {pontuacao} pontos.")

    # Fase 4
    fase_4(nome, pontuacao)
    print(f"\nObrigado por jogar, {nome}!")


iniciar_jogo()
```
